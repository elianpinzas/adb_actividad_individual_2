## Descripción general

El dataset posee información sobre alumnos en dos cursos dentro de dos escuelas en Portugal. La idea del trabajo presentado por Cortez y Silva era tratar de predecir, a partir de los datos recabados, la performance de los alumnos. El período estudiado es 2005-2006

### Contenido del archivo zip descargado
- **student-mat.csv**: datos de los alumnos del curso de matemática, 395 filas
- **student-por.csv**: datos de los alumnos del curso de lenguaje portugués, 649 filas
- **student.txt:** descripción de las columnas de los datasets. Adicionalmente posee información acerca de los alumnos en común entre ambos datasets
- **student-merge.R**: comandos en R para mostrar los alumnos en común entre ambos datasets

### Transcripción de la nota adjunta a los datasets (contenida en el archivo student.txt)
> _Additional note: there are several (382) students that belong to both datasets. These students can be identified by searching for identical attributes that characterize each student, as shown in the annexed R file._

__Considerando la nota adjunta a los datasets que indica que hay estudiantes que pertenecen a ambos datasets (382 en total) y para evitar tener que descartar duplicados u operaciones de mezcla de datasets, el EDA va a realizarse sobre el dataset más grande (student-por.csv)__

```python
path_portuguese_dataset="./dataset/student-por.csv"

df = pd.read_csv(path_portuguese_dataset, sep=";")
df.head()
```

```python
# Asignación de datos a las variables
dtype_map = {
    'school' : 'category',
    'sex' : 'category',
    'age' : 'float64',
    'address' : 'category',
    'famsize' : 'category',
    'Pstatus' : 'category',
    'Medu' : CategoricalDtype(categories=[0, 1, 2, 3, 4], ordered=True),
    'Fedu' : CategoricalDtype(categories=[0, 1, 2, 3, 4], ordered=True),
    'Mjob' : 'category',
    'Fjob' : 'category',
    'reason' : 'category',
    'guardian' : 'category',
    'traveltime' : CategoricalDtype(categories=[1, 2, 3, 4], ordered=True),
    'studytime' : CategoricalDtype(categories=[1, 2, 3, 4], ordered=True),
    'failures' : CategoricalDtype(categories=[0, 1, 2, 3, 4], ordered=True),
    'schoolsup' : 'category',
    'famsup' : 'category',
    'paid' : 'category',
    'activities' : 'category',
    'nursery' : 'category',
    'higher' : 'category',
    'internet' : 'category',
    'romantic' : 'category',
    'famrel' : CategoricalDtype(categories=[1, 2, 3, 4, 5], ordered=True),
    'freetime' : CategoricalDtype(categories=[1, 2, 3, 4, 5], ordered=True),
    'goout' : CategoricalDtype(categories=[1, 2, 3, 4, 5], ordered=True),
    'Dalc' : CategoricalDtype(categories=[1, 2, 3, 4, 5], ordered=True),
    'Walc' : CategoricalDtype(categories=[1, 2, 3, 4, 5], ordered=True),
    'health' : CategoricalDtype(categories=[1, 2, 3, 4, 5], ordered=True),
    'absences' : 'float64',
    'G1' : 'float64',
    'G2' : 'float64',
    'G3' : 'float64',
}

df = df.astype(dtype_map)

df.dtypes
```

```python
df.describe()
```

```python
# Código tomado de clase_02_intro_EDA_Titanic.ipynb
def cuartiles(variable):
    Q1 = variable.quantile(0.25)
    Q2 = variable.quantile(0.50)  # Percentile 5-0 - Equivalente a la mediana
    Q3 = variable.quantile(0.75)
    return Q1, Q2, Q3
```

```python
print("Datos en relación a la edad de la población estudiada")
age_mean = df['age'].mean()
age_median = df['age'].median()
age_mode = df['age'].mode()[0]

print(f"Media: {age_mean:.2f}")
print(f"Mediana: {age_median}")
print(f"Moda: {age_mode}")


fig, ax = plt.subplots(1, 2, figsize=(15, 4))

# Histograma
ax[0].hist(df['age'], bins=7)
ax[0].set_title('Histograma para edad')
ax[0].set_xlabel('Edad')
ax[0].set_ylabel('Alumnos')
ax[0].axvline(age_mean, color='black', ls='--', lw=2, label='Media')
ax[0].axvline(age_mode, color='red', ls='--', lw=3, label='Moda')
ax[0].axvline(age_median, color='yellow', ls='--', lw=2, label='Mediana')
ax[0].legend()

# Boxplot
sns.boxplot(data=df, x='age', ax=ax[1])

age_Q1, age_Q2, age_Q3 = cuartiles(df['age'])

age_IQR = age_Q3 - age_Q1
print(f"Cuartiles:")
print("Q1 (25%):", age_Q1)
print("Q2 (Mediana, 50%):", age_Q2)
print("Q3 (75%):", age_Q3)
print(f"Rango intercuartil (IQR): {age_IQR}")

lower_limit = age_Q1 - 1.5 * age_IQR
upper_limit = age_Q3 + 1.5 * age_IQR
ax[1].set_title('Boxplot para edad')
ax[1].grid(axis='y', ls='--')
ax[1].axvline(age_Q1,      color='blue',   linestyle='--', linewidth=1, label=f'Q1:         {age_Q1:.1f}')
ax[1].axvline(age_Q2,      color='yellow',    linestyle='--', linewidth=1, label=f'Mediana:  {age_Q2:.1f}')
ax[1].axvline(age_Q3,      color='green',  linestyle='--', linewidth=1, label=f'Q3:         {age_Q3:.1f}')
ax[1].axvline(lower_limit, color='purple', linestyle=':',  linewidth=1, label=f'Lím. inf:  {lower_limit:.1f}')
ax[1].axvline(upper_limit, color='brown',  linestyle=':',  linewidth=1, label=f'Lím. sup: {upper_limit:.1f}')
ax[1].legend()


plt.show()
```

```python
# Código tomado de clase_02_intro_EDA_Titanic.ipynb
def plot_pie(data, column, ax=None, startangle=90, palette='pastel', title=None):
    counts = data[column].value_counts()
    
    if ax is None:
        fig, ax = plt.subplots(figsize=(5, 3))
    
    ax.pie(
        counts,
        labels=counts.index,
        autopct='%1.1f%%',
        startangle=startangle,
        colors=sns.color_palette(palette)
    )
    ax.set_title(title if title else f'Distribución de "{column}"')
    ax.axis('equal') # Asegura que el gráfico sea circular
```

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.flatten() 

plot_pie(df, 'sex', ax=axes[0], title="Distribución de sexo de los estudiantes")
plot_pie(df, 'address', ax=axes[1], title="Zona de la vivienda (R: rural, U: urbana, variable 'address')")
plot_pie(df, 'activities', ax=axes[2], title="Realiza actividades extracurriculares, variable 'activities'")
plot_pie(df, 'nursery', ax=axes[3], title="Asistió a jardín, variable 'nursery'")
plot_pie(df, 'higher', ax=axes[4], title="Desea recibir educación superior, variable 'higher'")
plot_pie(df, 'internet', ax=axes[5], title="Tiene acceso a internet")

plt.tight_layout()
plt.show()
```

## Relación entre variables

```python
num_cols    = ['G1', 'G2', 'G3']
binary_cols = ['schoolsup', 'famsup', 'paid', 'activities', 'nursery', 'higher', 'internet', 'romantic']

pb_mat = pd.DataFrame(index=num_cols, columns=binary_cols, dtype=float)

for num_col in num_cols:
    for bin_col in binary_cols:
        mask = df[num_col].notna() & df[bin_col].notna()
        num  = df.loc[mask, num_col]
        bn   = df.loc[mask, bin_col].cat.codes   # 0/1
        r, p = pointbiserialr(bn, num) # pointbiserialr Devuelve r y p-value, pero solo nos interesa r
        pb_mat.loc[num_col, bin_col] = r

fig, ax = plt.subplots(figsize=(15, 5))
sns.heatmap(pb_mat.astype(float), annot=True, fmt=".2f", cmap="coolwarm",
            vmin=-1, vmax=1, linewidths=0.5,
            center=0, ax=ax)

ax.set_title("Punto Biserial", fontsize=11)
ax.set_xlabel("Var. binarias")
ax.set_ylabel("Var. numéricas")
plt.tight_layout()
plt.show()
```
