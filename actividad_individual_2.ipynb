## Descripción general

- El dataset posee información sobre alumnos en dos cursos dentro de dos escuelas en Portugal. La idea del trabajo presentado por Cortez y Silva era tratar de predecir, a partir de los datos recabados, la performance de los alumnos. El período estudiado es 2005-2006
- Se trata de dos archivos:
    - student-mat.csv (datos de los alumnos del curso de matemática, 395 filas)
    - student-por.csv (datos de los alumnos del curso de lenguaje portugués, 649 filas)

__Considerando la nota adjunta a los datasets que indica que hay estudiantes que pertenecen a ambos datasets (382 en total) y para evitar tener que descartar duplicados u operaciones de mezcla de datasets, el EDA va a realizarse sobre el dataset más grande (student-por.csv)__

## Tipos de variables

| Variable          | Descripción                           | Tipo      | Valores posibles  |
|-------------------|---------------------------------------|-----------|-------------------|
| `school`          | Escuela donde asiste                  | Categórica nominal (Binaria)   | "GP" (Gabriel Pereira) o "MS" (Mousinho da Silveira)|
| `sex`             | Sexo del estudiante                   | Categórica nominal (Binaria)   | "F" (femenino) o "M" (masculino)|
| `age`             | Edad del estudiante                   | Numérica discreta  | 15 a 22 |
| `address`         | Zona donde vive                       | Categórica nominal (Binaria)   | "U" (urbana) o "R" (rural) |
| `famsize`         | Tamaño de la familia                  | Categórica nominal (Binaria)   | "LE3" (<= 3) o "GT3" (>= 3)|
| `Pstatus`         | Cohabitación de los padres            | Categórica nominal (Binaria)   | "T" (viven juntos) o "A" (separados) |
| `Medu`            | Educación de la madre                 | Categórica ordinal  | 0: no posee, 1: ed. primaria hasta 4°, 2: ed. pr. (5° a 9°), 3: Secundaria, 4: Superior|
| `Fedu`            | Educación del padre                   | Categórica ordinal  | 0: no posee, 1: ed. primaria hasta 4°, 2: ed. pr. (5° a 9°), 3: Secundaria, 4: Superior|
| `Mjob`            | Trabajo de la madre                   | Categórica nominal   | "teacher", "health", "services", "at_home" or "other"|
| `Fjob`            | Trabajo del padre                     | Categórica nominal   | "teacher", "health", "services", "at_home" or "other"|
| `reason`          | Razón para elegir el colegio          | Categórica nominal   | "home" (es cerca de casa), "reputation", "course" (preferencia de curso), "other"|
| `guardian`        | Tutor del estudiante                  | Categórica nominal   | "mother", "father", "other" |
| `traveltime`      | Tiempo de viaje a la escuela          | Categórica ordinal  |  1 (<15 min), 2 (15-30 min), 3 (30 min-1 hora), 4  (>1 hora)|
| `studytime`       | Horas semanales de estudio            | Categórica ordinal  |  1 (<2 horas), 2 (2-5 horas), 3 (5-10 horas), 4 (>10 horas)|
| `failures`        | Número de clases desaprobadas         | Categórica ordinal  |  n si 1 <= n < 3, o 4 si es mayor a 3 |
| `schoolsup`       | Recibe soporte educacional extra      | Categórica nominal (Binaria)   | "yes" o "no" |
| `famsup`          | Recibe soporte educacional familiar   | Categórica nominal (Binaria)   | "yes" o "no" |
| `paid`            | Recibe clases extra pagas             | Categórica nominal (Binaria)   | "yes" o "no" |
| `activities`      | Realiza actividades extracurriculares | Categórica nominal (Binaria)   | "yes" o "no" |
| `nursery`         | Cursó jardín de infantes              | Categórica nominal (Binaria)   | "yes" o "no" |
| `higher`          | Quiere recibir educación superior     | Categórica nominal (Binaria)   | "yes" o "no" |
| `internet`        | Tiene internet en casa                | Categórica nominal (Binaria)   | "yes" o "no" |
| `romantic`        | Tiene una relación romántica          | Categórica nominal (Binaria)   | "yes" o "no" |
| `famrel`          | Calidad de relaciones familiares      | Categórica ordinal  |  De 1 a 5 (1: muy malo, 5: excelente) |
| `freetime`        | Tiene tiempo libre                    | Categórica ordinal  |  De 1 a 5 (1: muy malo, 5: excelente) |
| `goout`           | Sale con amigos                       | Categórica ordinal  |  De 1 a 5 (1: muy bajo, 5: muy alto) |
| `Dalc`            | Consume alcohol en la semana          | Categórica ordinal  |  De 1 a 5 (1: muy bajo, 5: muy alto) |
| `Walc`            | Consume alcohol en el fin de semana   | Categórica ordinal  |  De 1 a 5 (1: muy bajo, 5: muy alto) |
| `health`          | Estado de salud actual                | Categórica ordinal  |  De 1 a 5 (1: muy bajo, 5: muy alto) |
| `absences`        | Cantidad de ausentes                  | Numérica discreta |  De 0 a 93 |
| `G1`              | Nota del primer período               | Numérica discreta |  De 0 a 20 |
| `G2`              | Nota del segundo período              | Numérica discreta |  De 0 a 20 |
| `G3`              | Nota del tercer período               | Numérica discreta |  De 0 a 20 |
